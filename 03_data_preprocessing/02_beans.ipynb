## Data Splitting

Imputation and scaling both learn from values from the data. If we learn those values from the whole dataset, information about test rows leaks into the training process and our evaluation would be optimistic
#
So the process is:
1. Split into training and test(e.g: 70/30, stratified on target so both sets keep the same proportion, and a fixed random_state)
2. Learn every value from training only
3. apply the same values to the test

Stratification:
Bean A: 50
Bean B: 20
Bean C: 30
    Training        Test
A   50              0
B   20              0
C   0               30

this is bad because the model hasnt learned what Bean C is so that is why we should us stratification so that we keep the same ratio

```python
from sklearn.model_selection import train_test_split

X = beans.drop(columns='Class')# class is the target
y = beans['Class']

X_train, X_test, y_train, y_test = train_test_split(
    X,y,
    test_size=0.3,
    stratify=y,
    random_state=42
)

print(f"Training set: {X_train.shape}. Test set: {X_test.shape}")
print(f"Missing values in Training: {X_train.isnull().sum().sum()}, in test set: {X_test.isnull().sum().sum()}")
```
