## Data type Transformation

### Convert to boolean and Datetime

Always create a copy of the original dataframe for reproducibilty

```python
calendar = df.copy()
# str -> boolean 

calendar['available'] = df['available'].map({
    't': True,
    'f': False
})

# str -> datetime
calendar['date'] = pd.to_datetime(df['date'])
calendar.info()
```

```python
print(f"The period that date covers is from {calendar['date'].min()} to {calendar['date'].max()}")
```

```python
#df['date'].unique()
print(f"Days covered: {(calendar['date'].max() - calendar['date'].min()).days}")
```

```python
print(f" Available:{ {calendar['date'].mean()}}")
```
