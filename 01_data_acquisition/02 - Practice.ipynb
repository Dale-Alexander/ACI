```python
import requests
import os.path as osp
from pathlib import Path
import pandas as pd
from io import BytesIO
import zipfile 
from IPython.display import display
```

```python
url = 'https://archive.ics.uci.edu/static/public/275/bike+sharing+dataset.zip'
folder = Path('bike_practice')
folder.mkdir(exist_ok=True)
```

```python
response = requests.get(url, timeout=30)
print(f"Status Code : {response.status_code}")
print(f"Number of bytes: {len(response.content)}")
print(f"Content Type: {response.headers.get('Content-Type')}")
```

```python
with(zipfile.ZipFile(BytesIO(response.content)) as archive):
    print(f"Files: {archive.namelist()}")
    with(archive.open('day.csv') as csv_file):
        df = pd.read_csv(csv_file)##df is not closed to this scope, it is accessible from anywhere in this file
```

```python
path = folder / 'bikes.json'
df.to_json(path, orient='records')
df = pd.read_json(path, orient='records', convert_dates=False)
df.head()
#more than 5000 rentals, show dates, counts and how many days matched
busy_days = df.loc[df['cnt'] > 5000, ['dteday', 'cnt']]
print(len(busy_days))
```

```python
"""
Find the five working days with the highest rental counts. Show only `dteday`, `temp` and `cnt`. 
Save the result as `top_working_days.csv`, without an index.
"""

path = folder / 'top_working_days.csv'
high_rentals = df.loc[df['workingday'] == 1, ['dteday', 'temp', 'cnt']].sort_values('cnt', ascending=False).head(5)
high_rentals.to_csv(path, index=False)
```

```python
#Compare the mean, minimum and maximum daily rentals for working days and non-working days. Which group has the higher average?

#Try it below, then check the answer.
comparison = df.groupby('workingday')['cnt'].agg(
    mean='mean',
    max='max',
    min='min'
)
comparison

```

```python
#Create `casual_pct`: casual rentals as a percentage of total rentals. Show its first five values.
#Check whether `casual + registered` equals `cnt` for every day.

#Try it below, then check the answer.
df.head()
df['casual_pct'] = df['casual'] / (df['registered'] + df['casual']) * 100
print(df['casual_pct'].head())

match = (df['casual'] + df['registered'] == df['cnt']).all()
print(match)
```

```python
#Convert `dteday` to dates. Calculate total rentals for each year-month, keeping 2011 and 2012 separate. Show the three busiest months.

#Try it below, then check the answer.

df['dteday'] = pd.to_datetime(df['dteday'])
grouped = df.groupby(df['dteday'].dt.to_period('M'))['cnt']

# See the dictionary of {Period: index_positions}
print(grouped.apply(list))

high_rentals = df.groupby(df['dteday'].dt.to_period('M'))['cnt'].sum().sort_values(ascending=False).head(3)
high_rentals
```
