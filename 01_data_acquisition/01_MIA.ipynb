## Split Files (Data and Headers)

```python
# Read headers using standard file reading
path = osp.join('..', '..', 'data', 'mia', 'miaPassengers.names')

# open the file read only 'r' with specific encoding
with open(path, 'r', encoding='utf-8-sig') as f:
    # remove any whitespace, and split data on commas
    columns = f.read().strip().split(',')

path = osp.join('..', '..', 'data', 'mia', 'miaPassengers.data')
# data does not have a header, but we supply our own columns
df = pd.read_csv(path, header=None, names=columns)
df.head(3)
```
