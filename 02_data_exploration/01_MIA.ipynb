## Loading MIA Data

## Boxplot

From a quick look we notice that year is a sequential number (no need to analyse). 

Passengers variable appears to be large in numbers, hence we can create a visual to help us understand its distribution and general statistics.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 4))
plt.boxplot(df['passengers'], orientation='horizontal')
plt.title('Boxplot for Passengers')
plt.xlabel('Passengers')
plt.yticks([])
plt.show()
```

## Histograms and Scatterplots

These can be generated automatically for all column in dataset (also against each other) using seaborn pairplot.

```python
import seaborn as sns

sns.pairplot(df)
plt.show()
```

## Line Chart

A line chart is more suitable for a time-series chart. In this case we can see how passengers changed over the years.

```python
plt.figure(figsize=(8,6))
plt.plot(df['year'], df['passengers'])
plt.title('Passengers over the Years')
plt.xlabel('Year')
plt.ylabel('Passengers')
plt.xticks(df['year'], rotation=45)
plt.grid(True)
plt.show()
```
